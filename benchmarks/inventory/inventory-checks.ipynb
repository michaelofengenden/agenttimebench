One row per allocated task. This notebook checks source inventory integrity only. It never starts benchmark or model runs. Run from the inventory directory with the local cache available.

```python
from pathlib import Path
import json, importlib.util
folder = Path.cwd()
if not (folder / "tasks.json").exists():
    folder = folder / "benchmarks/inventory"
spec = importlib.util.spec_from_file_location("inventory_verify", folder / "verify_inventory.py")
module = importlib.util.module_from_spec(spec)
spec.loader.exec_module(module)
result = module.verify(folder.resolve().parents[1], hashes=True)
result
```
